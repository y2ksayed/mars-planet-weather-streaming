# Mars Weather — Gold Layer Exploration

Quick look at the daily aggregates produced by `src/streaming_job.py`.
Run the streaming job first so `data/output/gold` exists.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the gold layer written by the Spark job
gold = pd.read_parquet("../data/output/gold")
print(gold.shape)
gold.head()
```

```python
# Average temperature per sol
ax = gold.sort_values("sol").plot(x="sol", y="avg_temp_c", kind="line", marker="o")
ax.set_title("Avg Air Temperature per Sol (°C)")
ax.set_xlabel("Sol")
ax.set_ylabel("Temp °C")
plt.show()
```

```python
# Pressure vs wind — do dustier (windier) sols run lower pressure?
print(gold[["avg_pressure_pa", "avg_wind_ms"]].describe())
gold.plot.scatter(x="avg_wind_ms", y="avg_pressure_pa")
plt.show()
```
